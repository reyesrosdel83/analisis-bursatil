**Objetivo:** Automatizar la descarga de precios de cierre de acciones tecnológicas (MSFT, AAPL, NVDA) y generar un reporte en Excel para análisis posterior.

## 1. Importación de Librerías
Librerías necesarias para la extracción y manipulación de datos.

```python
import yfinance
```

## 2. Definición de Parámetros
Selección de activos y rango de fechas para el análisis.

```python
tickers=["MSFT", "AAPL", "NVDA"]
```

```python
data=yfinance.download(tickers,period="1mo")

cierres=data["Close"]

```

```python
cierres.to_excel("precios_cierres.xlsx")

print("¡Listo! Archivo creado: precios_cierres.xlsx")
print(cierres.tail())
```
